```python
try:
    number = 10
    result = number / 0
    print(result)

except ZeroDivisionError:
    print("0 দিয়ে ভাগ করা যায় না")
```

```python
count = 0
total = 0
max_age = 0
min_age = float("inf")

with open("students.txt", "r") as file:
    for line in file:
        name, age = line.strip().split(", ")
        age = int(age)

        count = count + 1
        total = total + age

        if age > max_age:
            max_age = age

        if age < min_age:
            min_age = age

average = total / count

print("Total Students:", count)
print("Total Age:", total)
print("Average Age:", average)
print("Maximum Age:", max_age)
print("Minimum Age:", min_age)
```
